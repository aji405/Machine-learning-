```python
# FIND-S Algorithm

data = [
    ['Sunny', 'Warm', 'Normal', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Yes'],
    ['Rainy', 'Cold', 'High', 'No'],
    ['Sunny', 'Warm', 'High', 'Yes']
]

h = data[0][:-1]

for row in data[1:]:
    if row[-1] == 'Yes':
        for i in range(len(h)):
            if h[i] != row[i]:
                h[i] = '?'

print("Final Hypothesis:", h)
```
